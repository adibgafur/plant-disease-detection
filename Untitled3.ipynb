## 2. Data Preprocessing for CNN

### Import TensorFlow

```python
import tensorflow as tf
```

### Create Image Data Generators

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

### Prepare the Training Data Generator

```python
train_datagen = ImageDataGenerator(
    rescale=1./255 # Normalize pixel values to [0, 1]
)
```

### Load the Training Images using `flow_from_directory`

```python
print(f"Loading training data from: {train_path}")
train_data = train_datagen.flow_from_directory(
    train_path,
    target_size=(128, 128), # Resize all images to 128x128 pixels
    batch_size=32, # Number of images per batch
    class_mode="categorical" # For multi-class classification
)
```

### Load the Validation Images using `flow_from_directory`

```python
val_datagen = ImageDataGenerator(
    rescale=1./255
)

print(f"Loading validation data from: {val_path}")
val_data = val_datagen.flow_from_directory(
    val_path,
    target_size=(128, 128),
    batch_size=32,
    class_mode="categorical",
    shuffle=False # Do not shuffle validation data for consistent evaluation
)
```

### Check the Class Labels Mapping

```python
print(train_data.class_indices)
```

### Get One Batch of Images and Labels

```python
images, labels = next(train_data)

print("Images shape:", images.shape)
print("Labels shape:", labels.shape)
```

### Visualize a Batch of Images

```python
plt.figure(figsize=(12, 8))

for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(images[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

### Understand the Labels

```python
print("First image label (one-hot encoded):")
print(labels[0]) # Display the first label from the batch

print("\nShape of the labels array:")
print(labels.shape)
```

### Check the Label of the First Image in the Batch

```python
class_indices = train_data.class_indices

index_to_class = {
    index: class_name
    for class_name, index in class_indices.items()
}

first_label_index = np.argmax(labels[0]) # Get the index of the highest probability for the first image
first_class_name = index_to_class[first_label_index]

print("First image in the batch belongs to:", first_class_name)
```

```python
plt.figure(figsize=(5, 5))

plt.imshow(images[0]) # Display the first image from the batch
plt.title(first_class_name)
plt.axis("off")

plt.show()
```

## 3. Build and Train the CNN Model

### Import Keras Modules for CNN

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

### Create the CNN Model Architecture

```python
num_classes = len(classes) # Get the total number of classes dynamically

model = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(128, 128, 3)),
    MaxPooling2D(2, 2),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D(2, 2),

    Conv2D(128, (3, 3), activation="relu"),
    MaxPooling2D(2, 2),

    Flatten(),

    Dense(128, activation="relu"),

    Dense(num_classes, activation="softmax") # Output layer with number of classes and softmax activation
])
```

### View the CNN Model Structure

```python
model.summary()
```

### Compile the CNN Model

```python
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

### Train the CNN Model

```python
epochs = 3 # You can increase this for better performance
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=epochs
)
```

## 4. Evaluate the Model

### Understand the Training Results History

```python
print(history.history.keys())
```

### Plot Training and Validation Accuracy

```python
plt.figure(figsize=(10, 5))

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training vs Validation Accuracy")
plt.legend()

plt.show()
```

### Plot Training and Validation Loss

```python
plt.figure(figsize=(10, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()

plt.show()
```

### Evaluate the Model on Validation Data

```python
val_loss, val_accuracy = model.evaluate(val_data)

print("Validation Loss:", val_loss)
print("Validation Accuracy:", val_accuracy)
```

### Get Model Predictions on Validation Data

```python
predictions = model.predict(val_data)

print("Prediction shape:", predictions.shape)
```

### Convert Predictions into Class Numbers

```python
predicted_classes = np.argmax(predictions, axis=1)

print("First 10 predicted class numbers:")
print(predicted_classes[:10])
```

### Convert Class Numbers to Class Names

```python
predicted_class_names = [
    index_to_class[index]
    for index in predicted_classes
]

print("First 10 predicted class names:")
print(predicted_class_names[:10])
```

### Get the Actual Labels from Validation Data

```python
actual_classes = val_data.classes

print("First 10 actual class numbers:")
print(actual_classes[:10])
```

### Compare Predicted vs Actual Labels (Overall Accuracy)

```python
correct_predictions = np.sum(predicted_classes == actual_classes)
total_predictions = len(actual_classes)

print("Correct predictions:", correct_predictions)
print("Total predictions:", total_predictions)

print(
    "Prediction Accuracy:",
    correct_predictions / total_predictions
)
```

### Create the Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(actual_classes, predicted_classes)

print("Confusion Matrix Shape:", cm.shape)
```

### Visualize the Confusion Matrix

```python
import seaborn as sns

plt.figure(figsize=(16, 14))

sns.heatmap(
    cm,
    cmap="Blues",
    xticklabels=classes, # Use class names for x-axis labels
    yticklabels=classes  # Use class names for y-axis labels
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Confusion Matrix")

plt.xticks(rotation=90)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Generate Classification Report

```python
from sklearn.metrics import classification_report

report = classification_report(
    actual_classes,
    predicted_classes,
    target_names=classes
)

print(report)
```

## 5. Test with Individual Images

### Test the Model with One Image from a Validation Batch

```python
# Re-get one batch of images and labels from validation data (if 'images' variable was overwritten)
# It's safer to get a fresh batch from val_data if train_data was used for 'images' earlier
# For consistency with the provided code, we'll assume 'images' here refers to a single image
# or the batch previously loaded from train_data. Let's use images[0] for a single test.

test_image_single = images[0] # Get the first image from the batch obtained from train_data

# Add a batch dimension: (128, 128, 3) -> (1, 128, 128, 3)
test_image_single_expanded = np.expand_dims(test_image_single, axis=0)

# Make a prediction
prediction_single = model.predict(test_image_single_expanded)

# Find the class with the highest probability
predicted_index_single = np.argmax(prediction_single)

# Convert the number to the class name
predicted_class_single = index_to_class[predicted_index_single]

print("Predicted class:", predicted_class_single)
```

### Show Prediction with the Image

```python
plt.figure(figsize=(6, 6))

plt.imshow(test_image_single)
plt.title("Predicted: " + predicted_class_single)
plt.axis("off")

plt.show()
```

### Test with a Completely New Image (Requires `new_leaf.jpg` to be present)

```python
from tensorflow.keras.preprocessing import image

# NOTE: new_leaf.jpg was not found. Using an image from the validation dataset for demonstration.
# We will take the first image from the validation data for prediction.

# Get one batch of images from the validation data
val_images, val_labels = next(val_data)

# Select the first image from the batch
img = val_images[0] # This will be a preprocessed and normalized image array

# Add batch dimension if it's not already present (ImageDataGenerator usually handles this for flow_from_directory)
# For a single image, we need to add a batch dimension for model.predict()
# If 'img' is already (128, 128, 3), we need to expand_dims.
# val_images[0] is already (128, 128, 3), so we'll just use it directly, no need to load_img or img_to_array
img_array = np.expand_dims(img, axis=0)

prediction_new = model.predict(img_array)

predicted_index_new = np.argmax(prediction_new)

predicted_class_new = index_to_class[predicted_index_new]

confidence_new = prediction_new[0][predicted_index_new] # Get confidence for the predicted class

print("Predicted Disease:", predicted_class_new)
print("Confidence:", f"{confidence_new:.2%}")
```
